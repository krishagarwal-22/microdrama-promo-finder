```python
# 2) Upload the episode-*.txt scripts into a folder called scripts/
import os, pathlib
from google.colab import files
pathlib.Path('scripts').mkdir(exist_ok=True)
up = files.upload()
for name, data in up.items():
    fn = name if name.startswith('episode-') else f'episode-{name}'
    pathlib.Path('scripts', fn).write_bytes(data)
print('saved', len(up), 'files')
```

```python
# 3) Run the pipeline and show the ranked shortlist
from promo_finder import parse_dir, find_all, get_scorer
episodes = parse_dir('scripts')
promos = find_all(episodes, get_scorer('heuristic'), per_episode=2)
print(f'{len(episodes)} episodes -> {len(promos)} promos\n')
for i, p in enumerate(promos[:15], 1):
    d = p.to_dict()
    seg = '  +  '.join(f"Ep{s.episode} {s.to_dict()['start']}-{s.to_dict()['end']}" for s in p.segments)
    print(f"#{i}  {seg}  score {d['score']}")
    print(f"     hook : {d['hook']}")
    print(f"     cliff: {d['cliffhanger'][:90]}\n")
```
